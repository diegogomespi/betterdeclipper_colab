```python
#@title INICIAR BETTERDECLIPPER
REPO_URL = "https://github.com/diegogomespi/betterdeclipper_colab.git"
BRANCH = "main"
PROJECT_DIR = "/content/betterdeclipper_colab"

import os
import subprocess
import sys

os.makedirs(os.path.dirname(PROJECT_DIR), exist_ok=True)
if not os.path.exists(PROJECT_DIR):
    subprocess.check_call(["git", "clone", "--branch", BRANCH, "--depth", "1", REPO_URL, PROJECT_DIR])
else:
    subprocess.check_call(["git", "fetch", "origin", BRANCH, "--depth", "1"], cwd=PROJECT_DIR)
    subprocess.check_call(["git", "reset", "--hard", f"origin/{BRANCH}"], cwd=PROJECT_DIR)

if PROJECT_DIR not in sys.path:
    sys.path.insert(0, PROJECT_DIR)

import importlib
import colab_launcher
colab_launcher = importlib.reload(colab_launcher)
colab_launcher.run_from_colab(REPO_URL, branch=BRANCH, project_dir=PROJECT_DIR, share=True)
```
